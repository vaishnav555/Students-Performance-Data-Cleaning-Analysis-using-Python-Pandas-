```python
numeric_cols = student.select_dtypes(include=['int64', 'float64']).columns
for col in numeric_cols:
    student[col].fillna(student[col].median(), inplace=True)
```

```python
categorical_cols = student.select_dtypes(include=['object']).columns
for col in categorical_cols:
    student[col].fillna(student[col].mode()[0], inplace=True)
```

```python

student.isnull().sum()
```

```python
before = student.shape[0]
student = student.drop_duplicates()
after = student.shape[0]

print("Rows before:", before)
print("Rows after:", after)
print("Duplicates removed:", before - after)
```

```python
student['Student_ID'] = student['Student_ID'].astype(int)
student.dtypes
```

```python
student['Average_Score'] = student[['Math_Score','Reading_Score','Writing_Score']].mean(axis=1)

def performance(score):
    if score >= 75:
        return "High"
    elif score >= 50:
        return "Medium"
    else:
        return "Low"

student['Performance_Level'] = student['Average_Score'].apply(performance)
```

```python
student.to_csv("cleaned_data.csv", index=False)
```
